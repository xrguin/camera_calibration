# iPhone calibration comparison

## tl;dr

The calibrations are **not interchangeable**. The new model has about 4.3% lower focal length and 2.1° wider horizontal effective FOV. On the new checkerboard views, using the old calibration increases PnP reprojection RMS from about 0.227 px to 0.788 px (3.48×). Across the image, corresponding pixel rays differ by 1.12° median and 1.43° at p95.

## Context & Methods

Both archives are expected to use 1920×1080 images and OpenCV's rational distortion model. Direct coefficient differences are not treated as physical differences because the numerator and denominator coefficients are strongly correlated. The comparison therefore uses effective FOV, ray-angle differences, same-ray pixel shifts, checkerboard reprojection, and PnP pose changes.

### Key assumptions

- The saved checkerboard images belong to their adjacent calibration archive.
- The board is the same 10×8-square board used by the launcher.
- The stated 0.20 m square size is used only for metric PnP pose differences; it does not affect intrinsics.

## Data

```python
from pathlib import Path
from usb_cam.compare_calibrations import compare

OLD_CALIBRATION = Path('usb_cam/iphone13wide/calibration.npz')
NEW_CALIBRATION = Path('usb_cam/iphone16wide/calibration.npz')
OLD_VIEWS = Path('usb_cam/iphone13wide/calib_views')
NEW_VIEWS = Path('usb_cam/iphone16wide/calib_views')
```

## Results

```python
comparison = compare(
    OLD_CALIBRATION, NEW_CALIBRATION, OLD_VIEWS, NEW_VIEWS,
    cols=10, rows=8, square=0.20,
)
comparison
```

## Takeaways

- Use `usb_cam/iphone13wide/calibration.npz` for the default iPhone 13 Wide pipeline.
- Use `usb_cam/iphone16wide/calibration.npz` only when that physical camera is selected.
- Do not substitute the previous calibration: the same-ray projection difference is 25.5 px median and 38.0 px at p95.
- The new fit itself is healthy (about 0.226 px re-fit RMS), though its 33 offline-usable views cover slightly less radius than the old 64-view run.
